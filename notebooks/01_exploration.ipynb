```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/customer_reviews.csv')
print(df.shape)
df.head()
```

```python
# Label Distribution
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

df['label'].value_counts().plot(kind='bar', ax=axes[0], color=['#e74c3c','#3498db','#2ecc71'])
axes[0].set_title('Sentiment Distribution')
axes[0].set_xlabel('Sentiment Class')
axes[0].set_ylabel('Count')

df['text_length'] = df['review'].str.split().str.len()
df.groupby('label')['text_length'].mean().plot(kind='bar', ax=axes[1], color=['#e74c3c','#3498db','#2ecc71'])
axes[1].set_title('Average Review Length by Sentiment')
axes[1].set_xlabel('Sentiment Class')
axes[1].set_ylabel('Avg Word Count')

plt.tight_layout()
plt.savefig('../docs/eda_plots.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Training history visualization
import json

with open('../src/models/saved/training_results.json') as f:
    results = json.load(f)

history = results['history']
epochs = range(1, len(history['train_loss']) + 1)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(epochs, history['train_loss'], 'b-', label='Train Loss')
axes[0].plot(epochs, history['val_loss'], 'r-', label='Val Loss')
axes[0].set_title('Training & Validation Loss')
axes[0].legend()

axes[1].plot(epochs, history['train_acc'], 'b-', label='Train Accuracy')
axes[1].plot(epochs, history['val_acc'], 'r-', label='Val Accuracy')
axes[1].set_title('Training & Validation Accuracy')
axes[1].legend()

plt.tight_layout()
plt.savefig('../docs/training_curves.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"Best Val Accuracy: {results['best_val_accuracy']*100:.2f}%")
print(f"Test Accuracy: {results['test_accuracy']*100:.2f}%")
```
